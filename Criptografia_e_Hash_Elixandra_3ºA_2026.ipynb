###**Aula 3**

Nesta aula, aprendemos a organizar nosso código de criptografia através da modularização, utilizando funções. Vimos como transformar o bloco de instruções original em uma função chamada codifica, que recebe a senha como parâmetro e utiliza o comando return para devolver o resultado. Também exploramos o uso de ferramentas de inteligência artificial para auxiliar na estruturação e melhoria desse código, além de identificar a necessidade de tratar caracteres especiais, como números, que podem gerar erros no algoritmo atual.

Aprendemos a refinar nossa lógica de criptografia utilizando uma estrutura condicional para identificar letras minúsculas entre 'a' e 'z'. Utilizamos a função ord para calcular a posição desses caracteres no alfabeto e processá-los via dicionário, enquanto garantimos que caracteres fora desse intervalo, como números e letras maiúsculas, sejam mantidos na senha sem alterações. Por fim, encapsulamos todo esse processo em uma função chamada codifica.

````python
dicionario = "qwertyuiopasdfghjklzxcvbnm"
alfabeto = "abcdefghijklmnopqrstuvwxyz"

def codifica(senha):
    senha_criptografada = ""
    for letra in senha:
        if letra in alfabeto:
            posicao = ord(letra) - ord("a")
            senha_criptografada += dicionario[posicao]
        else:
            senha_criptografada += letra
    return senha_criptografada

senha = "michi"
print(f"A senha '{senha}' criptografada é: {codifica(senha)}")
````

**Hora do Desafio**

Você criou uma lógica para criptografar letras minúsculas e também garantir que outros caracteres, como símbolos ou letras fora do intervalo, fossem mantidos no resultado.
O seu desafio agora é: Melhorar a função codifica para que ela também consiga criptografar letras maiúsculas e números, sem alterar os caracteres especiais.
Para isso, será necessário pensar em novas condicionais dentro do for, tratando separadamente:
* letras minúsculas;
* letras maiúsculas;
* números;
* outros caracteres.

💡Dica: você pode usar o Gemini como apoio para pensar em novos intervalos de comparação, como fizemos com "a" <= letra <= "z"

````python
dicionario = "qwertyuiopasdfghjklzxcvbnm"
dicionario_maiusculo = dicionario.upper()
dicionario_numeros = "9876543210"

def codifica(senha):
    senha_criptografada = ""
    for letra in senha:
        if "a" <= letra <= "z":
            posicao = ord(letra) - ord("a")
            senha_criptografada += dicionario[posicao]
        elif "A" <= letra <= "Z":
            posicao = ord(letra) - ord("A")
            senha_criptografada += dicionario_maiusculo[posicao]
        elif "0" <= letra <= "9":
            posicao = ord(letra) - ord("0")
            senha_criptografada += dicionario_numeros[posicao]
        else:
            senha_criptografada += letra
    return senha_criptografada

senha = "Michi@2024!"
print(f"A senha '{senha}' criptografada é: {codifica(senha)}")
````

###**Aula 4**

Nesta aula, vimos que algoritmos simples de substituição são frágeis e reversíveis. Por isso, exploramos o conceito de hash, que transforma informações em um valor numérico de forma irreversível. Aprendemos a implementar uma função básica que utiliza os valores da tabela ASCII para gerar esse hash e discutimos como diferentes entradas podem resultar no mesmo valor, reforçando a importância de métodos mais seguros para proteger dados.

````python
def calcula_hash(senha):
    valor = 0
    for letra in senha:
        valor = valor + ord(letra)
    return valor

print(calcula_hash("michi"))
````

###**Aula 7**

Nesta aula, exploramos o conceito de ataques de força bruta, focando em como senhas comuns podem ser facilmente descobertas ao compararmos seus hashes. Aprendemos a utilizar a biblioteca itertools para realizar análises combinatórias, demonstrando o risco de utilizar sequências previsíveis e a importância de criar senhas únicas e seguras.

````python
import itertools

def calcula_hash(senha):
    valor = 0
    for letra in senha:
        valor = valor*31 + ord(letra)
        valor %= 2**32
    return f"0x{valor:08x}"

hash_cadastrado = calcula_hash("abc")

letras = "abcdefghijklmnopqrstuvwxyz"
for tamanho in range(1, 4):
    for combinacao in itertools.product(letras, repeat=tamanho):
        tentativa = "".join(combinacao)
        if calcula_hash(tentativa) == hash_cadastrado:
            print(f"Senha descoberta: {tentativa}")
````

###**Aula 8**
Nesta aula, concluímos o estudo sobre ataques de força bruta, aprendendo como utilizar a biblioteca Itertools para gerar combinações exaustivas de caracteres a partir de um dicionário definido. Vimos que, ao comparar hashes em vez de senhas diretamente, conseguimos testar todas as possibilidades até encontrar a correspondência correta. Por fim, reforçamos a importância de criar senhas longas e complexas, já que o tempo necessário para realizar esse ataque aumenta drasticamente conforme o tamanho e a diversidade de caracteres da senha, tornando o processo inviável na prática.

````python
import itertools
import time

def calcula_hash(senha):
    valor = 0
    for letra in senha:
        valor = valor*31 + ord(letra)
        valor %= 2**32
    return f"0x{valor:08x}"

hash_cadastrado = calcula_hash("ab1")

caracteres = "abcdefghijklmnopqrstuvwxyz0123456789"

inicio = time.time()

for tamanho in range(1, 4):
    for combinacao in itertools.product(caracteres, repeat=tamanho):
        tentativa = "".join(combinacao)
        if calcula_hash(tentativa) == hash_cadastrado:
            print(f"Senha descoberta: {tentativa}")

fim = time.time()
print(f"Tempo: {fim - inicio:.2f} segundos")
````

💡Dica 2: o itertools.product gera combinações em forma de tuplas. Tuplas são estruturas parecidas com listas, mas usam parênteses e guardam vários valores juntos, por exemplo:

💡Dica 3: para transformar essa tupla em uma palavra ("ab"), usamos o método .join(), assim:

Esse comando junta os caracteres em uma única string.





````python
import itertools

def calcula_hash(senha):
    valor = 0
    for letra in senha:
        valor = valor*31 + ord(letra)
        valor %= 2**32
    return f"0x{valor:08x}"

hash_cadastrado = calcula_hash("abc")
letras = "abcdef"

def forca_bruta(hash_alvo, letras, max_tamanho):
    tentativas = 0
    for tamanho in range(1, max_tamanho + 1):
        for combinacao in itertools.product(letras, repeat=tamanho):
            tentativa = "".join(combinacao)
            tentativas += 1
            print(f"Tentativa {tentativas}: {tentativa}")
            if calcula_hash(tentativa) == hash_alvo:
                print(f"Senha descoberta: {tentativa} (em {tentativas} tentativas)")
                return tentativa

forca_bruta(hash_cadastrado, letras, 3)
````
